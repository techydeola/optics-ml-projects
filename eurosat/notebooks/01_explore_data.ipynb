```python
from pathlib import Path
import numpy as np
import tifffile

root = Path.cwd().parents[1] / "data" / "EuroSAT_MS"
classes = sorted(p.name for p in root.iterdir() if p.is_dir())
print(len(classes), classes)

for c in classes:
    print(f"{c:22s}", len(list((root / c).glob('*.tif'))))

f = next((root / classes[0]).glob("*.tif"))
img = tifffile.imread(f)
print(f.name, img.shape, img.dtype)
```

```python
for b in range(img.shape[-1] if img.shape[-1] == 13 else img.shape[0]):
    band = img[..., b] if img.shape[-1] == 13 else img[b]
    print(b, band.min(), band.max(), round(float(band.mean()), 1))
```

```python
import random
import pandas as pd

rng = random.Random(0)
means = {}
for c in ["Forest", "SeaLake", "Industrial", "River"]:
    files = sorted((root / c).glob("*.tif"))
    sample = rng.sample(files, 100)
    arr = np.stack([tifffile.imread(f) for f in sample]).astype(np.float64)  # (100, 64, 64, 13)
    means[c] = arr.mean(axis=(0, 1, 2))

df = pd.DataFrame(means).round(0)
df.index.name = "band_idx"
print(df)
```
